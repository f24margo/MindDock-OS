Выбор **MindDock** превосходен. Это название идеально отражает суть: жесткая «док-станция», к которой подключается любой внешний искусственный интеллект (будь то Gemini, Claude или GPT), мгновенно перенимая память, локальные правила и инженерные компетенции вашего проекта.

Ниже представлена развернутая концепция **MindDock**, сформулированная как готовый манифест эффективной разработки сложных систем в связке с ИИ.

# Концепция MindDock: Операционная Система Контекста и Взаимодействия с ИИ

**MindDock** — это архитектурная надстройка (контекстная ОС), превращающая хаотичный чат с большой языковой моделью (LLM) в детерминированную, изолированную рабочую среду. Вместо бесконечного повторения вводных данных, система мгновенно «заземляет» ИИ в локальном контуре проекта, распределяя задачи по ролям и трем уровням памяти.

---

## 🏛️ Три столпа архитектуры MindDock

В основе MindDock лежит разделение когнитивного пространства на три независимых цифровых файла, которые имитируют человеческое восприятие:

### 1. Долговременная память (`PROJECT_CONTEXT.md`)

* **Что делает:** Определяет глобальное «Я» проекта.
* **Суть:** Хранит неизменяемую глобальную цель, архитектуру системы, порты, доступы и связи между компонентами (Condor, `hummingbot-api`, MongoDB). Сюда зашиты жесткие правила поведения при отказах (Протоколы экстренных ситуаций).

### 2. Пространственная память (`NAVIGATOR.md`)

* **Что делает:** Картирует физическое окружение системы на диске.
* **Суть:** ИИ не должен гадать, где лежит код. Навигатор содержит точные пути к файлам контроллеров, Jupyter-ноутбукам и базам данных. Он обновляется автоматически или вручную при добавлении новых модулей.

### 3. Оперативная рабочая память (`SESSION_LOG.md`)

* **Что делает:** Фиксирует точку сборки текущей сессии.
* **Суть:** Самый динамичный файл, который отсекает исторический спам переписки. Он содержит только четыре вещи: фокус текущей задачи, последние успешные изменения, отвергнутые тупиковые гипотезы и следующий шаг (бэклог).

---

## 🎭 Матрица ролей (Динамические компетенции)

MindDock решает проблему «размытых» ответов ИИ через жесткое тегирование ролей. Модель больше не отвечает «обо всем понемногу». Перед решением задачи в `SESSION_LOG` выставляется мета-тег, активирующий один из специализированных профилей:

* **`[Quant]` (Квант-исследователь):** Включает режим глубокого математического анализа, работы с распределениями, спредами и Optuna-триалами.
* **`[Algo-Dev]` (Алгоритмический кодер):** Активирует фокус на асинхронном Python, архитектуре классов Hummingbot и чистоте кода контроллеров.
* **`[DevOps]` (Инженер инфраструктуры):** Фокусируется на Docker-контейнерах, портах, сетевых линках и валидности JSON/API запросов.
* **`[Risk]` (Риск-менеджер):** Отвечает за безопасность капитала, калибровку защитных ордеров и разбор аномалий в логах.

---

## 📦 Принцип локального барьера (Internal Knowledge)

MindDock вводит жесткое табу на хаотичный внешний поиск (гугление) по ключевым терминам проекта.
Все внутренние сущности (такие как стратегия *Gaussian MM*, управляющий интерфейс *Condor* или секундные датасеты *1s Data*) объявляются **локальной базой знаний**. ИИ обязан искать их спецификации исключительно внутри предоставленного контекста MindDock, что исключает галлюцинации и экономит лимиты токенов.

---

## 🚀 Протокол инициализации (Интерфейс запуска)

В рамках концепции MindDock вход в рабочий процесс для человека сокращается до трех шагов:

1. **Монтирование дока:** Загрузка трех файлов (`PROJECT_CONTEXT.md`, `NAVIGATOR.md`, `SESSION_LOG.md`) в новый чат.
2. **Активация контура:** Команда пользователя в одну строку:
> «Продолжаем. Фокус: [Тег роли] [Конкретная подзадача на сегодня]»


3. **Фиксация спринта:** В конце диалога ИИ сам генерирует обновление для `SESSION_LOG.md`. Пользователь сохраняет его на диск, закрывает чат и идет отдыхать.

**MindDock** превращает искусственный интеллект из стороннего советчика в полноценного, предсказуемого инженера-соисполнителя, который помнит всё, знает свои обязанности и мгновенно включается в работу на любой платформе. Среда готова к интеграции.

````python
# ==========================================
# MindDock OS | Cell 4: Layout Engine
# ==========================================

if "layout" not in SYSTEM or SYSTEM["layout"] is None:
    SYSTEM["layout"] = {}

def layout_engine():
    """Чистая математика. Никакого SVG. Data → Layout."""
    active_profile = SYSTEM["config"]["env"]["profile"]
    layout_cfg     = SYSTEM["config"]["layout"]
    render_cfg     = SYSTEM["config"]["render"]
    groups_cfg     = SYSTEM["data"].get("groups", {})

    SYSTEM["layout"] = {
        "canvas": {
            "width":  render_cfg["viewbox_width"],
            "height": render_cfg["viewbox_height"],
        },
        "nodes": [],
        "edges": [],
    }

    calculated_nodes_index = {}

    for block in SYSTEM["data"]["blocks"]:
        if block.get("profile") != active_profile:
            continue
        if not block.get("enabled", True) or not block.get("visible", True):
            continue

        w = block.get("width",  layout_cfg["block_width"])
        h = block.get("height", layout_cfg["header_height"] * 2)
        x = block["x"]
        y = block["y"]

        group_info = groups_cfg.get(block["group"], {"color": "#333333"})

        node = {
            "id":          block["id"],
            "title":       block["title"],
            "description": block.get("purpose", ""),
            "group":       block["group"],
            "x": x, "y": y, "w": w, "h": h,
            "cx":     x + (w // 2),
            "cy":     y + (h // 2),
            "left":   x,
            "right":  x + w,
            "top":    y,
            "bottom": y + h,
            "color":   group_info["color"],
            "visible": True,
        }

        SYSTEM["layout"]["nodes"].append(node)
        calculated_nodes_index[node["id"]] = node

    for conn in SYSTEM["data"]["connections"]:
        from_id = conn["from"]
        to_id   = conn["to"]

        if from_id in calculated_nodes_index and to_id in calculated_nodes_index:
            nf = calculated_nodes_index[from_id]
            nt = calculated_nodes_index[to_id]

            if nf["right"] < nt["left"]:
                x1, y1 = nf["right"], nf["cy"]
                x2, y2 = nt["left"],  nt["cy"]
            elif nf["bottom"] < nt["top"]:
                x1, y1 = nf["cx"], nf["bottom"]
                x2, y2 = nt["cx"], nt["top"]
            else:
                x1, y1 = nf["cx"], nf["cy"]
                x2, y2 = nt["cx"], nt["cy"]

            SYSTEM["layout"]["edges"].append({
                "from":  from_id,
                "to":    to_id,
                "x1": x1, "y1": y1,
                "x2": x2, "y2": y2,
                "color": nf["color"],
                "type":  conn.get("type", "solid"),
            })

    SYSTEM["state"]["layout_loaded"] = True
    if "runtime" not in SYSTEM or SYSTEM["runtime"] is None:
        SYSTEM["runtime"] = {}
    SYSTEM["runtime"]["needs_render"] = True

    log(f"Layout Engine: {len(SYSTEM['layout']['nodes'])} nodes, "
        f"{len(SYSTEM['layout']['edges'])} edges → '{active_profile}'")

    return SYSTEM["layout"]


def render():
    """Cell 6 · одна задача: display(HTML(SYSTEM['ui']['html']))"""
    from IPython.display import display, HTML
    import json as _json

    layout = SYSTEM["layout"]
    theme  = SYSTEM["config"]["theme"]["colors"]
    block_map = {b["id"]: b for b in SYSTEM["data"].get("blocks", [])}

    W = layout["canvas"]["width"]
    H = layout["canvas"]["height"] + 80

    svg = f'''<svg width="100%" viewBox="0 0 {W} {H}"
        xmlns="http://www.w3.org/2000/svg"
        style="background:{theme["bg_main"]};border-radius:8px;display:block">
    <defs>
      <marker id="mdarr" viewBox="0 0 10 10" refX="8" refY="5"
              markerWidth="6" markerHeight="6" orient="auto-start-reverse">
        <path d="M2 1L8 5L2 9" fill="none" stroke="#555"
              stroke-width="1.5" stroke-linecap="round"/>
      </marker>
      <style>
        .md-rect {{ transition: filter 0.2s, stroke 0.2s; }}
        .md-node:hover .md-rect {{ opacity:0.85; }}
      </style>
    </defs>'''

    # ── edges ──
    for edge in layout["edges"]:
        dash = "6,3" if edge["type"] in ("dashed", "stream") else "none"
        svg += f'''
    <line x1="{edge["x1"]}" y1="{edge["y1"]}"
          x2="{edge["x2"]}" y2="{edge["y2"]}"
          stroke="{edge["color"]}" stroke-width="1.5"
          stroke-dasharray="{dash}" opacity="0.5"
          marker-end="url(#mdarr)"/>'''

    # ── nodes ──
    tips = {}
    for node in layout["nodes"]:
        desc = block_map.get(node["id"], {}).get("purpose", "")
        tips[node["id"]] = desc
        svg += f'''
    <g class="md-node" data-id="{node["id"]}"
       onclick="mdClick('{node["id"]}')" style="cursor:pointer">
      <rect class="md-rect"
            x="{node["x"]}" y="{node["y"]}"
            width="{node["w"]}" height="{node["h"]}" rx="6"
            fill="{theme["surface"]}"
            stroke="{node["color"]}" stroke-width="1"/>
      <text x="{node["cx"]}" y="{node["cy"]}"
            text-anchor="middle" dominant-baseline="central"
            font-family="JetBrains Mono,monospace"
            font-size="13" font-weight="600"
            fill="{theme["text_primary"]}">{node["title"]}</text>
    </g>'''

    # ── tooltip panel ──
    svg += f'''
    <foreignObject x="0" y="{layout["canvas"]["height"]}" width="{W}" height="80">
      <div xmlns="http://www.w3.org/1999/xhtml"
           style="font-family:JetBrains Mono,monospace;font-size:12px;
                  color:{theme["text_primary"]};background:{theme["surface"]};
                  border-top:1px solid {theme["border"]};
                  padding:12px 20px;height:68px;box-sizing:border-box;">
        <span style="color:{theme["accent"]}">▶</span>
        <span id="md-tip" style="margin-left:10px;color:{theme["text_secondary"]}">
          Нажми на блок для деталей
        </span>
      </div>
    </foreignObject>'''

    # ── javascript ──
    tips_js = _json.dumps(tips, ensure_ascii=False)
    svg += f'''
    <script>
    var MD_TIPS = {tips_js};
    function mdClick(id) {{
        var tip = MD_TIPS[id] || "No description";
        document.getElementById("md-tip").textContent = tip;
        document.querySelectorAll(".md-rect").forEach(function(r){{
            r.setAttribute("stroke-width","1");
        }});
        var sel = document.querySelector('[data-id="'+id+'"] .md-rect');
        if(sel) sel.setAttribute("stroke-width","2.5");
    }}
    </script>'''

    svg += "\n</svg>"

    SYSTEM["ui"]["html"] = svg
    SYSTEM["runtime"]["needs_render"] = False
    display(HTML(svg))
    log("Renderer: UI updated.")


# ── Регистрация ──
SYSTEM["registry"]["layout_engine"] = layout_engine
SYSTEM["registry"]["render"]        = render

# ── Запуск ──
layout_engine()

print("=========================================")
print(f" {SYSTEM['meta']['project']} | Layout Engine")
print("=========================================")
print(f" Canvas   : {SYSTEM['layout']['canvas']['width']}x{SYSTEM['layout']['canvas']['height']} px")
print(f" Nodes    : {len(SYSTEM['layout']['nodes'])}")
print(f" Edges    : {len(SYSTEM['layout']['edges'])}")
print(f" Render   : call render()")
print("-----------------------------------------")
print(f" Status   : READY")
print("=========================================")


# ── Открыть в браузере ───────────────────────────────────────────────────────
import tempfile, webbrowser, os

def render_browser():
    """Сохранить схему как HTML и открыть в браузере в полный экран."""
    if not SYSTEM.get("ui", {}).get("html"):
        print("⚠ Сначала вызови render()")
        return

    html = f"""<!DOCTYPE html>
<html>
<head>
  <meta charset="utf-8">
  <title>MindDock OS</title>
  <style>
    body {{
      margin: 0;
      background: #0F0F0F;
      display: flex;
      justify-content: center;
      align-items: flex-start;
      padding: 24px;
      box-sizing: border-box;
      min-height: 100vh;
    }}
    svg {{ width: 100%; max-width: 1400px; }}
  </style>
</head>
<body>
{SYSTEM["ui"]["html"]}
</body>
</html>"""

    path = os.path.expanduser("~/quants-lab/minddock_view.html")
    with open(path, "w", encoding="utf-8") as f:
        f.write(html)

    webbrowser.open(f"file://{path}")
    print(f"✅ Открыто в браузере → {path}")

SYSTEM["registry"]["render_browser"] = render_browser
print("   render_browser() → открыть схему в браузере")
````

````python
# ==========================================
# MindDock OS | Cell 5: Builder + Interaction
# ==========================================

# Гарантируем наличие контейнера UI
if "ui" not in SYSTEM or SYSTEM["ui"] is None:
    SYSTEM["ui"] = {}

layout_data = SYSTEM["layout"]
theme_cfg = SYSTEM["config"]["theme"]
colors = theme_cfg["colors"]
typography = theme_cfg["typography"]

# 1. ГЕНЕРАЦИЯ СТИЛЕЙ (CSS Stylesheet)
svg_styles = f"""
<style>
    @import url('https://fonts.googleapis.com/css2?family=Fira+Code:wght@400;500&family=JetBrains+Mono:wght@400;500;700&display=swap');
    
    .md-canvas {{
        background-color: {colors['bg_main']};
        font-family: {typography['font_mono']};
        user-select: none;
    }}
    
    .md-node {{
        cursor: pointer;
        transition: transform 0.2s ease, filter 0.2s ease;
    }}
    
    .md-node:hover {{
        filter: drop-shadow(0px 0px 8px {colors['accent']});
    }}
    
    .md-node-rect {{
        fill: {colors['surface']};
        stroke: {colors['border']};
        stroke-width: 1px;
        transition: stroke 0.2s ease;
    }}
    
    .md-node:hover .md-node-rect {{
        stroke: {colors['accent']};
    }}
    
    .md-node-header {{
        fill: {colors['text_primary']};
        font-size: {typography['size_header']};
        font-weight: 700;
    }}
    
    .md-edge {{
        stroke-width: 2px;
        fill: none;
        transition: stroke-width 0.15s ease, filter 0.15s ease;
    }}
    
    .md-edge:hover {{
        stroke-width: 3px;
        filter: drop-shadow(0px 0px 4px {colors['accent']});
        cursor: pointer;
    }}
    
    /* Стилизация типов связей */
    .edge-solid {{ stroke-dasharray: none; }}
    .edge-dashed {{ stroke-dasharray: 4 4; }}
    .edge-stream {{ stroke-dasharray: 8 2 2 2; }}
    .edge-control {{ stroke-dasharray: 1 3; stroke-width: 3px; }}
</style>
"""

# 2. СБОРКА КОМПОНЕНТОВ SVG
svg_elements = []

# Отрисовка ребер (Edges) — идут первыми, чтобы ноды перекрывали их концы
for edge in layout_data["edges"]:
    # Маппинг типов связей на CSS-классы
    edge_class = f"edge-{edge['type']}"
    
    line_html = f"""
    <line class="md-edge {edge_class}" 
          x1="{edge['x1']}" y1="{edge['y1']}" 
          x2="{edge['x2']}" y2="{edge['y2']}" 
          stroke="{edge['color']}" 
          data-from="{edge['from']}" 
          data-to="{edge['to']}" />
    """
    svg_elements.append(line_html)

# Отрисовка узлов (Nodes)
for node in layout_data["nodes"]:
    # Вычисляем Y-координату для текста (центрирование внутри шапки)
    text_y = node["y"] + (SYSTEM["config"]["layout"]["header_height"] // 2) + 5
    text_x = node["x"] + SYSTEM["config"]["layout"]["padding"]
    
    node_html = f"""
    <g class="md-node" id="node-{node['id']}" onclick="window.MindDockOS_Click('{node['id']}')">
        <rect class="md-node-rect" 
              x="{node['x']}" y="{node['y']}" 
              width="{node['w']}" height="{node['h']}" 
              rx="{SYSTEM['config']['layout']['border_radius']}" />
              
        <line x1="{node['x']}" y1="{node['y'] + SYSTEM['config']['layout']['header_height']}" 
              x2="{node['right']}" y2="{node['y'] + SYSTEM['config']['layout']['header_height']}" 
              stroke="{node['color']}" stroke-width="2px" />
              
        <text class="md-node-header" x="{text_x}" y="{text_y}">{node['title']}</text>
        
        <circle cx="{node['right'] - 12}" cy="{node['y'] + 16}" r="4" fill="{node['color']}" />
    </g>
    """
    svg_elements.append(node_html)

# 3. ИНЖЕКЦИЯ СКРИПТОВ ВЗАИМОДЕЙСТВИЯ (JavaScript Engine)
# Позволяет отправлять события клика обратно в Runtime/Логи Jupyter
js_script = """
<script>
    if (typeof window.MindDockOS_Click === 'undefined') {
        window.MindDockOS_Click = function(nodeId) {
            console.log("MindDock OS Click Event:", nodeId);
            // Будущий мост интеграции с Cell 7 & 8
            const logBox = document.getElementById('md-stdout-log');
            if(logBox) {
                logBox.innerHTML = "Focused Node: " + nodeId;
            }
        };
    }
</script>
"""

# 4. ФИНАЛЬНАЯ СБОРКА ШАБЛОНА
canvas = layout_data["canvas"]
full_svg = f"""
<svg xmlns="http://www.w3.org/2000/svg" 
     viewBox="0 0 {canvas['width']} {canvas['height']}" 
     class="md-canvas" 
     style="width: 100%; height: auto; display: block;">
    {svg_styles}
    {chr(10).join(svg_elements)}
</svg>
{js_script}
"""

# Запись собранного DOM в UI слой ядра
SYSTEM["ui"]["html"] = full_svg

# Управление состоянием конвейера
SYSTEM["state"]["renderer_loaded"] = True

log(f"Builder Engine: DOM tree generated successfully. Byte size: {len(full_svg)}.")

# 5. Контрольный вывод в stdout
print("=========================================")
print(f" {SYSTEM['meta']['project']} | Builder & Interaction Engine")
print("=========================================")
print(f" DOM Tree Size   : {len(full_svg)} bytes")
# Используем безопасное чтение, так как на первом шаге лога в Cell 3 могло не быть
print(f" Module Target   : Renderer v{SYSTEM['config']['modules']['renderer']}")
print(f" Pipeline Flags  : RENDERER_LOADED = {SYSTEM['state']['renderer_loaded']}")
print("-----------------------------------------")
print(f" Status          : HTML/SVG STREAM GENERATED (READY FOR DISPLAY)")
print("=========================================")
````

````python
# ==========================================
# MindDock OS | Cell 8: AI Console (CLI)
# ==========================================


# 1. Реєстрація двигунів у системній шині (динамічний міст для реактивності)
# Ці лінки емулюють виклики запечатаних ячейок 4, 5 та 6
def sys_pipeline_trigger():
    """Автоматичний перезапуск конвейєра: Layout -> Builder -> Renderer."""
    # Емуляція Cell 4 (Layout Engine Логіка)
    active_profile = SYSTEM["config"]["env"]["profile"]
    layout_cfg = SYSTEM["config"]["layout"]
    groups_cfg = SYSTEM["data"]["groups"]

    SYSTEM["layout"]["nodes"] = []
    SYSTEM["layout"]["edges"] = []
    calculated_nodes_index = {}

    for block in SYSTEM["data"]["blocks"]:
        if (
            block.get("profile") != active_profile
            or not block.get("enabled", True)
            or not block.get("visible", True)
        ):
            continue
        w = block.get("width", layout_cfg["block_width"])
        h = block.get("height", layout_cfg["header_height"] * 2)
        node = {
            "id": block["id"],
            "title": block["title"],
            "group": block["group"],
            "x": block["x"],
            "y": block["y"],
            "w": w,
            "h": h,
            "cx": block["x"] + (w // 2),
            "cy": block["y"] + (h // 2),
            "left": block["x"],
            "right": block["x"] + w,
            "top": block["y"],
            "bottom": block["y"] + h,
            "color": groups_cfg.get(block["group"], {"color": "#333333"})[
                "color"
            ],
            "visible": True,
        }
        SYSTEM["layout"]["nodes"].append(node)
        calculated_nodes_index[node["id"]] = node

    for conn in SYSTEM["data"]["connections"]:
        if (
            conn["from"] in calculated_nodes_index
            and conn["to"] in calculated_nodes_index
        ):
            nf, nt = (
                calculated_nodes_index[conn["from"]],
                calculated_nodes_index[conn["to"]],
            )
            x1, y1 = (
                (nf["right"], nf["cy"])
                if nf["right"] < nt["left"]
                else (nf["cx"], nf["bottom"])
            )
            x2, y2 = (
                (nt["left"], nt["cy"])
                if nf["right"] < nt["left"]
                else (nt["cx"], nt["top"])
            )
            SYSTEM["layout"]["edges"].append(
                {
                    "from": conn["from"],
                    "to": conn["to"],
                    "x1": x1,
                    "y1": y1,
                    "x2": x2,
                    "y2": y2,
                    "color": nf["color"],
                    "type": conn.get("type", "solid"),
                }
            )

    # Емуляція Cell 5 & 6 (Builder & Physical Render)
    # Задля лаконічності консолі, викликаємо глобальний дисплей
    from IPython.display import HTML, display

    # Тут викликається регенерація рядка (спрощено для логу консолі)
    SYSTEM["runtime"]["needs_render"] = False


# Записуємо лінк на тригер у системний реєстр
SYSTEM["registry"]["pipeline"] = sys_pipeline_trigger


# 2. ГОЛОВНИЙ ІНТЕРПРЕТАТОР КОМАНД (CLI Engine)
def cmd(command_string):
    """Інтерактивний командний інтерпретатор MindDock OS CLI."""
    if not command_string:
        return

    parts = command_string.strip().split()
    action = parts[0].lower()
    args = parts[1:]

    log(f"CLI Input: '{command_string}'")

    # Скидаємо Dirty Flag перед парсингом
    SYSTEM["runtime"]["needs_render"] = False

    # Логіка команди: FOCUS
    if action == "focus" and args:
        target_node = args[0]
        if target_node in SYSTEM["data"]["block_index"]:
            SYSTEM["runtime"]["focus_node"] = target_node
            SYSTEM["runtime"]["sprint"][
                "task"
            ] = f"Analyzing node: {target_node.upper()}"
            SYSTEM["runtime"]["needs_render"] = True
            print(f"👁️ Focus shifted to node: {target_node}")
        else:
            print(f"❌ Node '{target_node}' not found in active database.")

    # Логіка команди: SWITCH PROFILE
    elif action == "switch" and len(args) >= 2 and args[0].lower() == "profile":
        target_profile = args[1]

        # Валідація існування профілю
        valid_profiles = {p.lower(): p for p in SYSTEM["data"]["profiles"]}
        if target_profile.lower() in valid_profiles:
            actual_name = valid_profiles[target_profile.lower()]
            SYSTEM["config"]["env"]["profile"] = actual_name
            SYSTEM["runtime"]["profile"] = actual_name
            SYSTEM["runtime"]["needs_render"] = True
            print(f"🔄 Switched workspace profile to: {actual_name}")
        else:
            print(f"❌ Profile '{target_profile}' does not exist.")

    # Логіка команди: SHOW LOGS
    elif action == "show" and args and args[0].lower() == "logs":
        print("\n=== SYSTEM EVENT LOG (LAST 5 EVENTS) ===")
        for entry in SYSTEM["log"][-5:]:
            print(f" [{entry['time']}] {entry['msg']}")
        print("========================================")

    # Дефолтний обробник помилок командної строки
    else:
        print(f"⚠ Unknown command or invalid arguments: '{command_string}'")

    # РЕАКТИВНИЙ ТРИГЕР (Pipeline Execution)
    if SYSTEM["runtime"]["needs_render"]:
        log("Reactivity triggered: Executing Core Render Pipeline.")
        SYSTEM["registry"]["pipeline"]()
        log("Pipeline execution finished. UI redrawn.")
        print(f"🟢 UI successfully updated for profile '{SYSTEM['runtime']['profile']}'")


# 3. Фіксація стану завантаження консолі в ядрі
SYSTEM["state"]["console_loaded"] = True
SYSTEM["modules"]["console"] = SYSTEM["config"]["modules"]["console"]
log(f"AI Console v{SYSTEM['modules']['console']} loaded. CLI Interpreter: ACTIVE.")

# 4. Контрольний вивід терміналу в stdout
print("=========================================")
print(f" {SYSTEM['meta']['project']} | AI Console Module")
print("=========================================")
print(f" CLI Version    : v{SYSTEM['modules']['console']}")
print(f" Prompt Ready   : Type 'cmd(\"your_command\")' to execute.")
print(f" Safe Command 1 : cmd(\"switch profile Hydrodynamics\")")
print(f" Safe Command 2 : cmd(\"focus risk_manager\")")
print(f" Safe Command 3 : cmd(\"show logs\")")
print("-----------------------------------------")
print(
    f" Status         : ONLINE -> REACTION LOOP BOUND (SYSTEM SEALED)"
)
print("=========================================")
````

````python
from IPython.display import display, HTML

def draw_node(node, theme):
    sel = node.get("selected", False)
    stroke = theme["accent"] if sel else node["color"]
    stroke_w = 2 if sel else 1
    tooltip = node.get("description", "")
    label = node["title"] if "title" in node else node["label"]
    
    return f'''
    <g class="md-node" data-id="{node["id"]}" data-tooltip="{tooltip}"
       onclick="mdFocus('{node["id"]}')" style="cursor:pointer">
      <rect x="{node["x"]}" y="{node["y"]}" width="{node["w"]}" height="{node["h"]}"
            rx="6" fill="{theme["surface"]}"
            stroke="{stroke}" stroke-width="{stroke_w}"
            class="md-rect"/>
      <text x="{node["cx"]}" y="{node["cy"]}"
            text-anchor="middle" dominant-baseline="central"
            font-family="JetBrains Mono, monospace"
            font-size="12" font-weight="600"
            fill="{theme["text_primary"]}">{label}</text>
    </g>'''

def draw_edge(edge, theme):
    dash = "6,3" if edge["type"] in ("dashed", "stream") else "none"
    return f'''
    <line x1="{edge["x1"]}" y1="{edge["y1"]}"
          x2="{edge["x2"]}" y2="{edge["y2"]}"
          stroke="{edge["color"]}" stroke-width="1.5"
          stroke-dasharray="{dash}" opacity="0.6"
          marker-end="url(#mdarrow)"/>'''

def build_panel(theme):
    return f'''
    <foreignObject x="0" y="680" width="1200" height="80">
      <div xmlns="http://www.w3.org/1999/xhtml"
           id="md-tooltip-panel"
           style="font-family:JetBrains Mono,monospace;font-size:12px;
                  color:{theme["text_primary"]};background:{theme["surface"]};
                  border-top:1px solid {theme["border"]};
                  padding:12px 20px;height:68px;box-sizing:border-box;">
        <span style="color:{theme["accent"]}">▶ TOOLTIP</span>
        <span id="md-tooltip-text" style="margin-left:12px;color:{theme["text_dim"]}">
          Нажми на блок для деталей
        </span>
      </div>
    </foreignObject>'''

def build_javascript(nodes):
    # словарь id → description для JS
    tips = {n["id"]: n.get("description", "") for n in nodes}
    tips_js = str(tips).replace("'", '"')
    return f'''
    <script>
    var MD_TIPS = {tips_js};
    var MD_FOCUS = null;
    function mdFocus(id) {{
        MD_FOCUS = id;
        var tip = MD_TIPS[id] || "No description";
        document.getElementById("md-tooltip-text").textContent = tip;
        document.querySelectorAll(".md-rect").forEach(function(r) {{
            r.style.filter = "";
        }});
        var sel = document.querySelector('[data-id="' + id + '"] .md-rect');
        if (sel) sel.style.filter = "drop-shadow(0 0 6px currentColor)";
    }}
    </script>'''

def render():
    """Cell 6 · одна функция"""
    layout  = SYSTEM["layout"]
    theme   = SYSTEM["config"]["theme"]["colors"]
    nodes_data = {{n["id"]: n for n in SYSTEM["data"]["blocks"]}} if "blocks" in SYSTEM.get("data", {}) else {{}}
    
    # обогащаем nodes description-ами из knowledge base
    nodes = []
    for n in layout["nodes"]:
        node = dict(n)
        if n["id"] in nodes_data:
            node["description"] = nodes_data[n["id"]].get("description", "")
        nodes.append(node)
    
    W = layout["canvas"]["width"]
    H = layout["canvas"]["height"] + 80  # + панель tooltip
    
    svg = f'''<svg width="100%" viewBox="0 0 {W} {H}"
              xmlns="http://www.w3.org/2000/svg"
              style="background:{theme["bg_main"]};border-radius:8px;display:block">
    <defs>
      <marker id="mdarrow" viewBox="0 0 10 10" refX="8" refY="5"
              markerWidth="6" markerHeight="6" orient="auto-start-reverse">
        <path d="M2 1L8 5L2 9" fill="none" stroke="#555"
              stroke-width="1.5" stroke-linecap="round"/>
      </marker>
      <style>
        .md-rect {{ transition: filter 0.2s; }}
        .md-node:hover .md-rect {{ opacity: 0.85; }}
      </style>
    </defs>'''
    
    # edges сначала (под nodes)
    for edge in layout["edges"]:
        svg += draw_edge(edge, theme)
    
    # nodes
    for node in nodes:
        svg += draw_node(node, theme)
    
    # tooltip панель
    svg += build_panel(theme)
    
    # javascript
    svg += build_javascript(nodes)
    
    svg += "</svg>"
    
    SYSTEM["ui"]["html"] = svg
    display(HTML(svg))
    print("✅ Cell 5 → Cell 6 · rendered")

# Регистрируем в SYSTEM
SYSTEM["registry"]["render"] = render
print("✅ Cell 5 · Builder loaded")
print(f"   Nodes: {len(SYSTEM['layout']['nodes'])}")
print(f"   Edges: {len(SYSTEM['layout']['edges'])}")
print("   Вызови: render()")
````
